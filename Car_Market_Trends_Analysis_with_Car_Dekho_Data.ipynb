```python
# Plot selling price over manufacturing year grouped by fuel type
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x='Year',
    y='Selling_Price',
    hue='Fuel_Type',
    size='Present_Price',
    sizes=(30, 250),
    alpha=0.8
)
plt.title('Used Car Selling Price vs. Manufacturing Year by Fuel Type', fontsize=13, fontweight='bold')
plt.xlabel('Manufacturing Year')
plt.ylabel('Selling Price (Lakh INR)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Plot average selling price across dealer vs individual sellers and transmission types
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x='Seller_Type',
    y='Selling_Price',
    hue='Transmission',
    palette='viridis',
    errorbar=None
)
plt.title('Average Selling Price by Seller Type and Transmission', fontsize=13, fontweight='bold')
plt.xlabel('Seller Channel')
plt.ylabel('Average Selling Price (Lakh INR)')
plt.tight_layout()
plt.show()
```

```python
# Function to calculate and print IQR bounds and outlier counts
def print_outlier_summary(df, column):
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    upper_bound = q3 + 1.5 * iqr
    outliers = df[df[column] > upper_bound]
    print(f"[{column}] Threshold: > {upper_bound:.2f} | Outliers Count: {len(outliers)}")
```

```python
# Display outlier counts for key numerical features
print("--- OUTLIER SUMMARY (1.5x IQR Rule) ---")
print_outlier_summary(df, 'Selling_Price')
print_outlier_summary(df, 'Present_Price')
print_outlier_summary(df, 'Kms_Driven')
```

```python
# Plot Boxplots to visualize outliers
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
sns.boxplot(ax=axes[0], y=df['Selling_Price'], color='skyblue')
axes[0].set_title('Selling Price Outliers (Lakh INR)', fontweight='bold')

sns.boxplot(ax=axes[1], y=df['Present_Price'], color='salmon')
axes[1].set_title('Present Price Outliers (Lakh INR)', fontweight='bold')

sns.boxplot(ax=axes[2], y=df['Kms_Driven'], color='lightgreen')
axes[2].set_title('Kilometers Driven Outliers', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Bivariate scatter plot: Kms_Driven vs. Selling_Price grouped by Fuel_Type
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Kms_Driven', y='Selling_Price', hue='Fuel_Type', alpha=0.8)
plt.title('Bivariate Analysis: Kilometers Driven vs. Selling Price', fontweight='bold')
plt.xlabel('Kilometers Driven')
plt.ylabel('Selling Price (Lakh INR)')
plt.tight_layout()
plt.show()
```
